```python
#estrazione dati da csv
from pathlib import Path
import pandas as pd

def extract_data(file_path):
    try:
        data = pd.read_csv(file_path)
        return data
    except FileNotFoundError:
        print(f"Error: The file at {file_path} was not found.")
        return None
    except pd.errors.EmptyDataError:
        print("Error: The file is empty.")
        return None
    except pd.errors.ParserError:
        print("Error: There was a parsing error while reading the file.")
        return None

filename = "Sales_transactions_2022_2025.csv"
working_dir = Path.cwd().resolve()
csv_path = next(
    (parent / "dataset" / filename
     for parent in (working_dir, *working_dir.parents)
     if (parent / "dataset" / filename).is_file()),
    None,
)

if csv_path is None:
    raise FileNotFoundError(f"Could not find dataset/{filename} from {working_dir}")

raw_df = extract_data(csv_path)
df = raw_df.copy()
df.head()
```

```python
df.info()
```

```python

#operazioni di pulizia e normalizzazione dei dati
#   convertire Order_Date e Order_Time in formato datetime      --FATTO
#   individuare eventuali valori anomali o outlier nelle colonne numeriche e gestirli di conseguenza    --FATTO
#   inserire i product_name mancanti in base al product_id      --FATTO
#   eliminare eventuali duplicati basati su Transaction_ID      --FATTO 
```

```python
#verifica dei duplicati basati su Transaction_ID
df[df['Transaction_ID'].duplicated()]
```

```python
#rimozione dei duplicati basati su Transaction_ID
df = df.drop_duplicates(subset='Transaction_ID', keep='first')
df[df['Transaction_ID'].duplicated()]
```

```python
#verifica dei product_name mancanti
df[df['Product_Name'].isnull()][['Product_ID', 'Product_Name']]
```

```python
# Verifica che ogni Product_ID abbia un solo nome noto
nomi_noti = df.dropna(subset=["Product_Name"])
conflitti = nomi_noti.groupby("Product_ID")["Product_Name"].nunique()
if (conflitti > 1).any():
    raise ValueError(f"Product_ID associati a più nomi: {conflitti[conflitti > 1].index.tolist()}")

# Crea la corrispondenza ID → nome e completa i valori mancanti
mappa_nomi = nomi_noti.drop_duplicates("Product_ID").set_index("Product_ID")["Product_Name"]
df["Product_Name"] = df["Product_Name"].fillna(df["Product_ID"].map(mappa_nomi))
```

```python
#ricerca di eventuali valori anomali ad esempio su valori percentuali
df[(df['Discount_Percentage'] > 100) | (df['Discount_Percentage'] < 0)]['Discount_Percentage']
```

```python
#eliminazione deivalori anomali su Discount_Percentage
df = df[(df['Discount_Percentage'] <= 100) & (df['Discount_Percentage'] >= 0)]
```

```python
#si nota come Customer_Segment cambi nel tempo - implementazione di SCD di tipo 2 per Customer_Segment
df[df['Customer_ID'] == 'C01591']
```

```python
#conversione delle colonne Order_Date e Order_Time in formato datetime
df["Order_Date"] = pd.to_datetime(
            df["Order_Date"], errors="raise"
        ).dt.date
df["Order_Time"] = pd.to_datetime(
            df["Order_Time"], format="%H:%M:%S", errors="raise"
        ).dt.time
```

```python
#renedere i nomi delle colonne in minuscolo per uniformità e facilità di accesso
df.columns = [column.lower() for column in df.columns]
df.head()
```

```python
#verifica di valori negativi nella colonna quantity
df[df['quantity'] < 0][['quantity', 'order_date', 'order_time', 'product_id', 'product_name']]
```

```python
#rimozioni dei valori negativi nella colonna quantity rendendoli positivi
df['quantity'] = df['quantity'].abs()
```
